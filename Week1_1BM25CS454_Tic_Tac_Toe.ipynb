```python
import random

def print_board(board):
    """Prints the current state of the game board."""
    print("\n")
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print("\n")

def check_winner(board, player_char):
    """Checks if a whole row, column, or diagonal is filled with the player's character."""
    win_conditions = [ [3, 4, 5], [6, 7, 8] ]
    for condition in win_conditions:
        if board[condition[0]] == board[condition[1]] == board[condition[2]] == player_char:
            return True
    return False

def is_board_full(board):
    """Checks if the board is completely full (resulting in a draw)."""
    return " " not in board

def get_computer_move(board, computer_char, human_char):
    """Determines the computer's move using rule-based logic instead of complex algorithms."""

    for i in range(9):
        if board[i] == " ":
            board_copy = board.copy()
            board_copy[i] = computer_char
            if check_winner(board_copy, computer_char):
                return i


    for i in range(9):
        if board[i] == " ":
            board_copy = board.copy()
            board_copy[i] = human_char
            if check_winner(board_copy, human_char):
                return i


    available_moves = [i for i, cell in enumerate(board) if cell == " "]
    return random.choice(available_moves)

def play_game():
    """Main function managing the game loop between HUMAN and COMPUTER."""
    board = [" " for _ in range(9)]

    print("--- TIC-TAC-TOE: HUMAN VS COMPUTER ---")


    human_char = ""
    while human_char not in ["X", "O"]:
        human_char = input("Choose your character (X or O): ").upper()

    computer_char = "O" if human_char == "X" else "X"
    print(f"HUMAN is '{human_char}' | COMPUTER is '{computer_char}'")


    current_turn = random.choice(["HUMAN", "COMPUTER"])
    print(f"\nThe game starts with: {current_turn}!")

    print("\nPositions are numbered 1 through 9:")
    print(" 1 | 2 | 3 \n---|---|---\n 4 | 5 | 6 \n---|---|---\n 7 | 8 | 9 ")


    while True:
        print_board(board)

        if current_turn == "HUMAN":

            try:
                move = int(input("HUMAN, choose an empty cell (1-9): ")) - 1
            except ValueError:
                print("Invalid input! Please enter a number.")
                continue

            if move < 0 or move > 8 or board[move] != " ":
                print("Invalid or occupied cell! Try again.")
                continue

            board[move] = human_char

            if check_winner(board, human_char):
                print_board(board)
                print("🎉 Game Over! HUMAN wins!")
                break

            current_turn = "COMPUTER"

        else:

            print("COMPUTER is making its move...")
            move = get_computer_move(board, computer_char, human_char)
            board[move] = computer_char
            print(f"COMPUTER placed '{computer_char}' at position {move + 1}")

            if check_winner(board, computer_char):
                print_board(board)
                print("💻 Game Over! COMPUTER wins!")
                break

            current_turn = "HUMAN"


        if is_board_full(board):
            print_board(board)
            print("🤝 Game Over! The game is a draw!")
            break

if __name__ == "__main__":
    play_game()
```

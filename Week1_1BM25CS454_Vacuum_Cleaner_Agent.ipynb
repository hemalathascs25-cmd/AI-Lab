```python


room_A = input("Is Room A dirty? (yes/no): ").lower()
room_B = input("Is Room B dirty? (yes/no): ").lower()
location = input("Enter vacuum cleaner location (A/B): ").upper()


room_A_dirty = room_A == "yes"
room_B_dirty = room_B == "yes"

print("\nInitial State:")
print("Room A:", "Dirty" if room_A_dirty else "Clean")
print("Room B:", "Dirty" if room_B_dirty else "Clean")
print("Vacuum Cleaner Location:", location)

print("\nActions:")


while room_A_dirty or room_B_dirty:


    if location == "A":

        if room_A_dirty:
            print("Vacuum Cleaner sucks dirt from Room A.")
            room_A_dirty = False
        else:
            print("Room A is already clean.")
            print("Vacuum Cleaner moves from Room A to Room B.")
            location = "B"


    elif location == "B":

        if room_B_dirty:
            print("Vacuum Cleaner sucks dirt from Room B.")
            room_B_dirty = False
        else:
            print("Room B is already clean.")
            print("Vacuum Cleaner moves from Room B to Room A.")
            location = "A"


print("\nGoal State Reached!")
print("Room A: Clean")
print("Room B: Clean")
print("Vacuum Cleaner Location:", location)
print("Both rooms are clean.")
```

```python

import random
import math

N = 8

def calculate_cost(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def generate_neighbor(board):
    neighbor = board.copy()

    col = random.randint(0, N - 1)
    new_row = random.randint(0, N - 1)

    while new_row == neighbor[col]:
        new_row = random.randint(0, N - 1)

    neighbor[col] = new_row
    return neighbor


def simulated_annealing():
    current = [random.randint(0, N - 1) for _ in range(N)]
    current_cost = calculate_cost(current)

    best = current.copy()
    best_cost = current_cost

    temperature = 10.0
    cooling_rate = 0.99
    max_iterations = 100000

    for iteration in range(max_iterations):
        if current_cost == 0:
            break

        if temperature < 0.0001:

            current = [random.randint(0, N - 1) for _ in range(N)]
            current_cost = calculate_cost(current)
            temperature = 10.0

        neighbor = generate_neighbor(current)
        neighbor_cost = calculate_cost(neighbor)

        delta = neighbor_cost - current_cost


        if delta < 0 or random.random() < math.exp(-delta / temperature):
            current = neighbor
            current_cost = neighbor_cost


        if current_cost < best_cost:
            best = current.copy()
            best_cost = current_cost

        temperature *= cooling_rate

    return best, best_cost


def display_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print("-", end=" ")
        print()



solution, cost = simulated_annealing()

print("Simulated Annealing - 8 Queens Problem")
print("Name: Hemalatha S USN: 1BM25CS454")
print("Solution:", solution)
print("Final Cost:", cost)

if cost == 0:
    print("\nSolution Found!")
    print("\nChessboard:")
    display_board(solution)
else:
    print("\nSolution not found. Please run again.")
```
